## Multi-Model Post-Fire Landslide Probability Workflow

This notebook is the reproducible execution layer for the post-fire Landlab workflow. The intent is that the upstream data pipeline prepares the required rasters and forcing files, and this notebook documents how those inputs are loaded, transformed, and passed into the hydrology and landslide-probability components.

Imported layers are described before the derived fields that depend on them. This keeps the sequence transparent for review, reruns, and future transfer to stakeholder-facing workflows.


### 1. Import Libraries and Components

This notebook keeps the scientific workflow visible, but the operational controls should live outside the notebook. Best practice is to store the input root path, layer file names, per-layer no-data values, forcing-file path, event window, and key model constants in a separate YAML or JSON config file, then read that config in one short setup cell.

Using a notebook cell for configuration is acceptable for quick tests, but a separate config file is the better long-term practice for reproducibility.

### 15. Create Leaf Area Index Field

Assign LAI values to plant functional types to create a derived vegetation field for radiation, PET, and interception-related calculations.


```python
LAI_grass = 1.5
LAI_shrub = 2.0
LAI_tree  = 4.0

## create or replace the field
grid.add_field('vegetation__live_leaf_area_index',
               np.zeros(grid.number_of_nodes, dtype=float),
               at='node', clobber=True)

pft = grid.at_node['vegetation__plant_functional_type']
lai = grid.at_node['vegetation__live_leaf_area_index']  # we can pass this directly

for i in range(grid.number_of_nodes):
    if pft[i] == 0:
        lai[i] = LAI_grass
    elif pft[i] == 1:
        lai[i] = LAI_shrub
    elif pft[i] == 2:
        lai[i] = LAI_tree
    elif pft[i] == 3:
        lai[i] = 1   ## default for bare

grid['cell']['vegetation__live_leaf_area_index']=map_node_to_cell(grid, "vegetation__live_leaf_area_index")
```

### 25. Run Daily Hydro-Eco Simulation

For each day, apply rainfall and temperature forcings, run the Radiation, PET, and SoilMoisture components, and store runoff and recharge outputs. This section is where rainfall is partitioned into interception, runoff, evapotranspiration, storage change, and root-zone leakage, so it directly controls the recharge signal later passed to the landslide model.

Key parameters to control in a production workflow include the forcing file, event window, input root path, per-layer no-data values, and the site constants used above.
